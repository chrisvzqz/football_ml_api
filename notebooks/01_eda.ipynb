```python
victorias = {}

for _, partido in df.iterrows():
    if partido['winner'] == 'HOME_TEAM':
        victorias[partido['home_team']] = victorias.get(partido['home_team'], 0) + 1

    elif partido['winner'] == 'AWAY_TEAM':
        victorias[partido['away_team']] = victorias.get(partido['away_team'], 0) + 1

equipo_mas_victorias = max(victorias, key=victorias.get)

print(f"Equipo con más victorias: {equipo_mas_victorias}")
print(f"Victorias: {victorias[equipo_mas_victorias]}")
```
