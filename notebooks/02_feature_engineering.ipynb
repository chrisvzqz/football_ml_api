## Función para aplicar a todos los partidos

```python
# Create home_points_last_5 y away_points_last_5

df_copy = df_new.copy()

def get_points_last_5(team, current_date, season):
    previous_matches = df_copy[((df_copy['home_team'] == team) | (df_copy['away_team'] == team)) & (df_copy['utc_date'] < current_date) & (df_copy['season_start_year'] == season)].sort_values(by='utc_date', ascending=False).head(5)

    puntos = 0
    for _, row in previous_matches.iterrows():
        if row['home_team'] == team:
            if row['winner'] == 'HOME_TEAM':
                puntos += 3
            elif row['winner'] == 'AWAY_TEAM':
                puntos += 0
            else:
                puntos += 1
        else:
            if row['winner'] == 'HOME_TEAM':
                puntos += 0
            elif row['winner'] == 'AWAY_TEAM':
                puntos += 3
            else:
                puntos += 1
    
    return puntos

df_copy['home_points_last_5'] = df_copy.apply(lambda row: get_points_last_5(row['home_team'], row['utc_date'], row['season_start_year']), axis=1)
df_copy['away_points_last_5'] = df_copy.apply(lambda row: get_points_last_5(row['away_team'], row['utc_date'], row['season_start_year']), axis=1)
```
